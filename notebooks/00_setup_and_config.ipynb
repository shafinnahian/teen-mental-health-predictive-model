```python
from __future__ import annotations

import os
import sys
from pathlib import Path
from typing import Optional

_colab_root = Path("/content/teen-mental-health-predictive-model")
if _colab_root.is_dir():
    os.chdir(_colab_root)


def find_project_root(start: Optional[Path] = None) -> Path:
    """Walk upward until requirements.txt and src/config.py are found."""
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "requirements.txt").exists() and (candidate / "src" / "config.py").exists():
            return candidate
    raise FileNotFoundError(
        "Could not locate project root. "
        "Expected a directory containing requirements.txt and src/config.py."
    )


PROJECT_ROOT = find_project_root()
print("Project root:", PROJECT_ROOT)

%pip install -q -r {PROJECT_ROOT / "requirements.txt"}
```
